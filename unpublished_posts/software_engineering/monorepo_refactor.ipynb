# Monorepo Restructure

Let's create a reference monorepo using turbo, so that we can reference its basic structure during the refactor. 

Run this command, it will guide you through the creation of the monorepo. Create a my-turborepo monorepo in the current directory: 

```bash
pnpm dlx create-turbo@latest
```

Since we're using `pnpm`, we first need to ensure we set up a `pnpm-workspace/yaml` as:
