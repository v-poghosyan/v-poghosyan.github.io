# Start with a floor plan

You can make one using some free browser-based tools, but also you can use Blender with [BonsaiBIM](https://docs.bonsaibim.org/) extension that allows you to treat Blender as a BIM tool (like AutoCAD) and manage IFC files (things like walls, doors, windows, and much more that are standard in the architectural industry). In theory, an architectural model is just a BIM, and a BIM is just a schema file with the location of all the IFC components. You can either use a BIM builder to create your floor plan, or work with the schema itself by editing the IFC file directly. 

## Install BonsaiBIM extension in Blender

In Blender Go to **Edit** > **Preferences** > **Get Extensions** and **Allow Online Access**, then search for "Bonsai" and install the extension. A new **BIM** tab should now show up at the top, and you can choose **File** > **New IFC Project** to start a new project. You can then use the tools in the BIM tab to create walls, doors, windows, and other architectural elements. You can also import an existing IFC file if you have one, or export your project as an IFC file to use in other software.

# Gather reference material and get inspired

Look for reference material and use [pureRef](https://www.pureref.com/) to organize it.

# Import floor plan into Blender

Create a plane in Blender, and then import the floor plan image as a texture. You can do this by going to the `Shading` workspace, creating a new material for the plane, and then adding an `Image Texture` node. Add a "human" model to the scene for scale reference, and then connect the `Image Texture` node to the `Base Color` input of the `Principled BSDF` shader. Scale as needed to match the "human" for scale. Remember the increments in Blender are $1$ meter by default, the "human" model could be anywhere between $1.6$ to $1.8$ meters, so scale the plane accordingly.
