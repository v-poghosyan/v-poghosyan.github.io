# Pencil drawing

I used the HB pencil brush in Procreate to put together a rough perspective drawing of the scene (although I didn't bother with perspective lines). I also drew the reflections inside the chrome candleholder in this pencil-drawing stage, because I realized that chrome is the absence of material information, so drawing the chrome candleholder is a matter of wrapping the visual information coming from the environment (my living room, the scene) onto the object. So, essentially you're drawing the scene again. We can see that even in this pencil-drawing phase (as opposed to the painting phase which comes later), the candleholder already reads somewhat like chrome.

# Broad strokes

Once we have the pencil sketch, we can approximate the colors. I essentially made a detailed palette of colors on the right and put them in their rough positions on the canvas (on the left).  

::: {#fig-color-palette}

![](./assets/Still_Life_(Color_Palette).png){width=700}

:::

This is a step beyond just laying down the base color of the objects. A lot of material information comes from the color information (the texture channel usually enhances the material, but the heavy lifting is done by color and shading). Some people like to do a grayscale painting first. It could be a good idea, but I skipped it this time. I just worked with the right colors from the get-go and tried to shade in such a way as to convey form and material information.

# Laying on the paint

This phase is all about smudging the colors from the palette we created above to block in the main color information of the objects. I should've broken this step down further, but there's no mystery here. Just hours of using the smudge tool, some form of solid brush, and some form of airbrush... The cyan cup has no texture information, so it already reads as its true material after this step. That is, the color information and the lighting information (especially the highlights) give away the entire object. Insofar as the bowl reads like wood, it's because of the motion of the brush strokes. I'll apply the wooden texture in a later phase.

# Texturing

The technique I like to use for this is two-pronged:

## 1. Make a clipping mask

1. Create a new layer.
2. Make the new layer a clipping mask.
3. Move the new layer above the layer that contains the color information (created above), i.e., the very same layer containing the object to be textured.
4. Paint the texture information (see next step) on this new clipping mask layer. You'll notice that you can't paint beyond the boundaries of your object on the layer below -- that's the beauty of clipping masks.

Here's how to do that.

::: {#fig-clipping-mask}

![](./assets/Clipping_mask.png){width=700}

Clipping mask goes on top. Paint the texture information on the clipping mask (not directly on the color layer itself). Then, merge down onto the color layer when satisfied with the results.  

:::


::: {.callout-tip title="Tip" appearance="minimal" collapse="false"}
We can stack multiple clipping masks on top of each other and then merge down as required. This is a great non-destructive workflow.
:::


## 2. Use a textured brush to achieve a texture

Textured brushes seem useless when applied by themselves, but the key is to apply them on their own layer and then distort that layer to match the perspective of the scene. In a sense, you're transforming the space of the new layer into the space of the layer below it (the layer containing the object being textured).

1. Select any brush with a texture that looks like the texture of the material you're trying to achieve. It's usually simple because material brushes are typically named after a material. Procreate comes with many already, but we can always download more from the community.
2. Create a new layer (above the colored object's layer, as before).
3. Draw with the material brush on this new layer.
4. Use Warp, Distort, or any other free-form transformation tool in Procreate's Selection mode to map the just-created texture onto the object -- see screenshot below.
5. Use the eraser (an airbrush eraser) and the smudge tool to massage the shape and position of the now-transformed texture.
6. (Optional) Create a clipping mask above the new texture layer to re-color the texture for any last-minute effects. This is always a non-destructive option for doing so and offers the benefit of only drawing on top of the texture below it.
7. When happy with the results, merge the texture layer down onto the color layer. If you performed the optional step 6, then merge the clipping mask onto the texture layer first; then merge that onto the color layer. You can still use the smudge tool at this point, but be aware that doing so will remove the new texture information. Sometimes that's just what we need, though.

::: {#fig-distort-tool}

![](./assets/Distort_tool.png){width=700}

Use the distort tool to map the texture to the space of the object (i.e. put the texture in perspective)

:::

# Define the background

Sometimes in drawing we are presented with a choice to do things in a certain logical order. In this case, I wanted to finish painting the chrome candleholder first, but I remembered that chrome is the absence of material and just the reflection of the environment in which the object lives. I decided to define a simple environment using a simple gradient. I didn't use the actual gradient tool, just a base color and an airbrush of a slightly lighter shade.


::: {#fig-background-defined}

![](./assets/Still_Life_(Background_Defined).png){width=700}

:::

Note that I also drew the base of the still life (the white stand that carries the objects). It was reflected in the chrome candleholder. I used the same clipping mask technique described above to draw the shadows of the base, the cup, and the candleholder, distinguishing between hard and soft shadow edges. 

Some rules for shadows:

1. The edge of the shadow is hard if the shadow is nearer the object; the edge softens the further it is from the object.
2. Objects leave dark shadows right below them (where they touch the base). This shadow is what makes objects look "grounded" rather than floating on top of the platform. It even has a name... ["ambient occlusion"](https://en.wikipedia.org/wiki/Ambient_occlusion)

I colored and textured the lemons, and just colored the bookcase and the books (coloring was enough, apparently, to make the IKEA KALLAX shelves look like they have a wood texture, plus it's fine to be a little lazy about the background). For the books, it was all about getting the colors right and the reflections on the shelf (the smudge tool helped here).

But, more importantly, with the background now fully defined, I could paint the reflective chrome candleholder. This was just a matter of filling in the warped lines I drew earlier (of the reflections on the candleholder) with color from the environment. A similar color palette technique could be used as in the **Broad strokes** section above. This is a recursive workflow, sort of -- if we recall that famous nested mirrors and reflections visual metaphor for recursion from computer science textbooks. But that's overkill as an actual strategy; it's more of a thought-provoking application that's fun to think about. We can apply the same principle to a smaller problem... But we can also certainly just eyeball it. I eyeballed it here.

And, even though chrome is the absence of material, the object itself could still have some smudges from fingertips of people handling it. Not to mention some distortions due to manufacturing. Procreate has good brushes for this sort of effect; I used a clipping mask layer to add a few smudges and fingerprints to the candleholder as a last touch. Airbrushing with a very light color, as if there were mist in the room, had a huge effect here.
